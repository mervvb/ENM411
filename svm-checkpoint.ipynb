```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

from sklearn.svm import SVR

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Dropout
```

```python
df = pd.read_excel("veri.xlsx")
df.head()
```

```python
df.shape
df.info()
```

```python
df.isnull().sum()
```

```python
# Unnamed kolonlarını daha anlamlı hale getirelim
df = df.rename(columns={
    "Unnamed: 0": "unit_number",
    "Unnamed: 1": "time_cycle",
    "Unnamed: 2": "operational_setting_1",
    "Unnamed: 3": "operational_setting_2",
    "Unnamed: 4": "operational_setting_3"
})

df.head()
```

```python
missing_table = pd.DataFrame({
    "Eksik Sayısı": df.isnull().sum(),
    "Eksik Oranı (%)": df.isnull().mean() * 100
})

missing_table
```

```python
df.describe().T
```

```python
constant_cols = []

for col in df.columns:
    if df[col].nunique() <= 1:
        constant_cols.append(col)

constant_cols
```

```python
df_clean = df.drop(columns=constant_cols)

df_clean.shape
```

```python
corr = df_clean.corr()

plt.figure(figsize=(16, 12))
sns.heatmap(corr, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
target = "Low-pressure turbines Cool air flow"
```

```python
X = df_clean.drop(columns=[target])
y = df_clean[target]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
split_index = int(len(df_clean) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print(X_train.shape, X_test.shape)
print(y_train.shape, y_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print("Train veri:", X_train_scaled.shape)
print("Test veri:", X_test_scaled.shape)
print("Hedef değişken:", target)
```

```python
###
```

```python
svm = SVR(kernel='rbf', C=10, epsilon=0.1)

svm.fit(X_train_scaled, y_train)

y_pred_svm = svm.predict(X_test_scaled)
```

```python
rmse_svm = np.sqrt(mean_squared_error(y_test, y_pred_svm))
mae_svm = mean_absolute_error(y_test, y_pred_svm)
r2_svm = r2_score(y_test, y_pred_svm)

print("SVM RMSE:", rmse_svm)
print("SVM MAE:", mae_svm)
print("SVM R2:", r2_svm)
```

```python
plt.figure(figsize=(10,5))
plt.plot(y_test.values[:200], label="Gerçek")
plt.plot(y_pred_svm[:200], label="Tahmin")
plt.legend()
plt.title("SVM Tahmin vs Gerçek")
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "C": [1, 10, 50],
    "epsilon": [0.01, 0.1, 0.5],
    "kernel": ["rbf"]
}

grid = GridSearchCV(SVR(), param_grid, cv=3, scoring="r2")
grid.fit(X_train_scaled, y_train)

print("En iyi parametreler:", grid.best_params_)
```

```python
best_svm = SVR(C=1, epsilon=0.1, kernel='rbf')

best_svm.fit(X_train_scaled, y_train)

y_pred_svm_best = best_svm.predict(X_test_scaled)
```

```python
rmse_svm_best = np.sqrt(mean_squared_error(y_test, y_pred_svm_best))
mae_svm_best = mean_absolute_error(y_test, y_pred_svm_best)
r2_svm_best = r2_score(y_test, y_pred_svm_best)

print("Best SVM RMSE:", rmse_svm_best)
print("Best SVM MAE:", mae_svm_best)
print("Best SVM R2:", r2_svm_best)
```

```python
plt.figure(figsize=(10,5))
plt.plot(y_test.values[:200], label="Gerçek")
plt.plot(y_pred_svm_best[:200], label="Best SVM Tahmin")
plt.legend()
plt.title("Best SVM Tahmin vs Gerçek")
plt.show()
```

```python
def create_sequences(X, y, window_size=10):
    X_seq = []
    y_seq = []

    for i in range(len(X) - window_size):
        X_seq.append(X[i:i+window_size])
        y_seq.append(y.iloc[i+window_size])

    return np.array(X_seq), np.array(y_seq)
```

```python
window_size = 10

X_seq, y_seq = create_sequences(X_train_scaled, y_train, window_size)
X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test, window_size)

print(X_seq.shape)
print(y_seq.shape)
```

```python
model = Sequential()

model.add(GRU(64, return_sequences=True, input_shape=(window_size, X_train.shape[1])))
model.add(Dropout(0.2))

model.add(GRU(32))
model.add(Dropout(0.2))

model.add(Dense(1))

model.compile(optimizer='adam', loss='mse')
```

```python
history = model.fit(
    X_seq, y_seq,
    epochs=20,
    batch_size=32,
    validation_split=0.2
)
```

```python
y_pred_gru = model.predict(X_test_seq)
```

```python
rmse_gru = np.sqrt(mean_squared_error(y_test_seq, y_pred_gru))
mae_gru = mean_absolute_error(y_test_seq, y_pred_gru)
r2_gru = r2_score(y_test_seq, y_pred_gru)

print("GRU RMSE:", rmse_gru)
print("GRU MAE:", mae_gru)
print("GRU R2:", r2_gru)
```
